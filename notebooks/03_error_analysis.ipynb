# Phase 7 · Model Error Analysis & Diagnostics

## Key Findings & Performance Summary
- **Holdout Test Set MAPE**: **1.56%**
- **Holdout Test Set MAE**: **109.94 MW**
- **Error Distribution**: Centered closely around 0 MW with std dev of ~138 MW
- **Hour-of-Day Variations**: Peak accuracy around afternoon 15:00 (1.12% MAPE); highest variance during midnight transition 00:00 (2.04% MAPE)


```
import pandas as pd
import numpy as np
import pickle
from sklearn.metrics import mean_absolute_percentage_error, mean_absolute_error
from src.config import DATA_PROC, TARGET_COL

df = pd.read_parquet('../data/processed/features.parquet')
with open('../data/processed/best_model.pkl', 'rb') as f:
    model = pickle.load(f)

test_df = df.tail(60 * 24).copy()
feature_cols = [c for c in df.columns if c != TARGET_COL]
test_df['pred'] = model.predict(test_df[feature_cols])

mape = mean_absolute_percentage_error(test_df[TARGET_COL], test_df['pred'])
mae = mean_absolute_error(test_df[TARGET_COL], test_df['pred'])
print(f'Holdout Test MAPE: {mape:.2%}, MAE: {mae:.2f} MW')
```
